```python
import random, time
from route_planner import RoadNetwork

def random_city(n, seed=7):
    rnd = random.Random(seed)
    g = RoadNetwork()
    names = [f"N{i}" for i in range(n)]
    for i in range(1, n):
        g.add_road(names[i], names[rnd.randrange(i)], rnd.randint(1, 20))
    extra = n // 2
    for _ in range(extra):
        u, v = rnd.sample(names, 2)
        if v not in g.adj[u]:
            g.add_road(u, v, rnd.randint(1, 20))
    return g

for n in (10, 25, 50, 100):
    g = random_city(n)
    t = time.perf_counter(); bfs(g, "N0"); tb = time.perf_counter() - t
    t = time.perf_counter(); dijkstra(g, "N0", f"N{n-1}"); td = time.perf_counter() - t
    print(f"n={n} roads={len(g.roads())} bfs={tb:.6f}s dijkstra={td:.6f}s")
```
